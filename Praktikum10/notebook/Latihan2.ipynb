```python
import pandas as pd
from sklearn.metrics import confusion_matrix, accuracy_score, precision_score, recall_score
```

```python
df = pd.read_excel("/content/drive/MyDrive/Mecine Learning/Praktikum10/data/data_prediksi.xlsx")
print("Data Prediksi:")
display(df)
```

# Confusion Matrix

```python
cm = confusion_matrix(df["Hasil Sebenarnya"], df["Hasil Prediksi"], labels=["Lulus", "Tidak Lulus"])

cm_df = pd.DataFrame(cm,
                     index=["Actual Lulus", "Actual Tidak Lulus"],
                     columns=["Pred Lulus", "Pred Tidak Lulus"])

print("\nConfusion Matrix:")
display(cm_df)
```

## Hitung Accuracy, Precision, Recall

```python
accuracy = accuracy_score(df["Hasil Sebenarnya"], df["Hasil Prediksi"])
precision = precision_score(df["Hasil Sebenarnya"], df["Hasil Prediksi"], pos_label="Lulus")
recall = recall_score(df["Hasil Sebenarnya"], df["Hasil Prediksi"], pos_label="Lulus")

print("Accuracy:", round(accuracy * 100, 2), "%")
print("Precision:", round(precision * 100, 2), "%")
print("Recall:", round(recall * 100, 2), "%")
```
